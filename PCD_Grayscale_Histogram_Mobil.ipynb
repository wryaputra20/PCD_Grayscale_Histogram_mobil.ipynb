```python

import imageio.v2 as imageio
import numpy as np
import matplotlib.pyplot as plt
from google.colab import files

# 1. Membaca citra
uploaded = files.upload()
nama_file = next(iter(uploaded))

img = imageio.imread(nama_file)
img = img[:, :, :3]  # Mengambil 3 kanal RGB

print("Citra RGB:", img.shape)

# 2. Mengubah citra RGB ke grayscale
R = img[:, :, 0]
G = img[:, :, 1]
B = img[:, :, 2]

img_gray = (
    0.299 * R +
    0.587 * G +
    0.114 * B
).astype(np.uint8)

print("Citra Grayscale:", img_gray.shape)

# 3. Menghitung jumlah piksel
tinggi, lebar = img_gray.shape
jumlah_piksel = img_gray.size

print("Tinggi:", tinggi, "Piksel")
print("Lebar:", lebar, "Piksel")
print("Jumlah:", jumlah_piksel, "Piksel")

# 4. Menghitung histogram
histogram = np.bincount(
    img_gray.flatten(),
    minlength=256
)

print("Total piksel histogram:", histogram.sum())

# 5. Menghitung jumlah piksel setiap intensitas
for intensitas in range(256):
    print(
        f"Intensitas {intensitas:3d}: "
        f"{histogram[intensitas]} piksel"
    )

# 6. Mencari intensitas dominan
intensitas_dominan = np.argmax(histogram)
jumlah_intensitas = histogram[intensitas_dominan]

print("Intensitas Dominan:", intensitas_dominan)
print("Jumlah Intensitas:", jumlah_intensitas)

# 7. Menampilkan gambar grayscale dan histogram
plt.figure(figsize=(14, 5))

plt.subplot(1, 2, 1)
plt.imshow(img_gray, cmap="gray")
plt.title("Citra Grayscale")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.bar(range(256), histogram, width=1)
plt.title("Histogram Grayscale")
plt.ylabel("Jumlah Piksel")
plt.xlabel("Nilai Intensitas")
plt.xlim(0, 255)

plt.tight_layout()
plt.show()

print("TINGGI:", img_gray.shape[0], "piksel")
print("LEBAR:", img_gray.shape[1], "piksel")
print("TOTAL PIKSEL:", img_gray.size)
print("TOTAL HISTOGRAM:", histogram.sum())
```
